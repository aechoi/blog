# Interactive Kalman Filter

Investigate each step and component of the Kalman Filter. The
measurement model is the identity (**C = I**), so measurements can be
depicted in the state space. A variety of dynamic models are available
in the dropdown menu. Control is omitted for simplicity (**B = 0**).

**Solid** outlines and **filled** dots are things you can change;
**dashed** outlines and **hollow** dots are derived by the filter. Drag
a square handle to reshape a covariance, hover anything to see what it
is, drag empty space to pan, scroll or pinch to zoom. The **stage**
buttons highlight the predict or update step you are looking at, and the
legend entries below the plot toggle any entity on its own. Every
ellipse is a 1σ contour.
