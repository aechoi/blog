# HRTF Spatialization

A browser-based follow-up to [my earlier post on Head Related Transfer
Functions](../blog/head_related_tfs.qmd). An audio can be processed to
sound like it’s coming from a specific point in space around your head.
This version runs entirely client-side, so performance may vary
depending on your hardware.

**Headphones required** — binaural spatialization depends on each ear
hearing an independent signal, which speakers can’t give you.

Works best in Chrome or Edge. Firefox and safari work too, but can be
slower. Use the ⛶ button in the corner to enter full screen mode.
