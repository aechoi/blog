# Conductors and Capacitance

An interactive electrostatics solver for a cross-section of long
conductors and insulators. Conductors can be held at a fixed potential
or fixed charge (floating). Insulators can be assigned a dielectric and
charge density. Field lines, equipotential lines, and charge
distributions can be displayed. Selecting a conductor shows its
capacitance: total capacitance, capacitance to the conductor at
infinity, and mutual capacitance to every other conductor.
