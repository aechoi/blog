# Electromagnetic Field Visualizer

A real-time 3D visualization of the electric and magnetic fields of a
moving positive point charge, computed directly from the
Lienard-Wiechert potentials. Retardation is exact: field points see the
charge’s position, velocity, and acceleration at the retarded time found
by solving the light-cone equation against the charge’s recorded
trajectory, so kinks in the field visibly propagate outward at speed $c$
whenever the charge suddenly accelerates.

Pick a preset motion (circular or oscillating, around any axis) to see
radiation patterns, or let it meander (a random walk splined through
waypoints confined to a ball around the origin). You can also drag the
charge directly with your mouse (confined to the basis plane of
whichever slice orientation is selected). The crimson arrow at the
charge shows its instantaneous acceleration.

Drag on empty space to orbit the camera or enable “Auto orbit”. Change
the view to “2D” to see the waves that travel along 1 dimension of the
axes.
