# Free and Bound Charge

An interactive simulation of discrete charges in conductors and
insulators. Place fixed and free charges, draw conducting or insulating
regions, and watch electrons rearrange: free charge in a conductor moves
to its surfaces until the field inside vanishes, while bound charge in
an insulator only shifts slightly, polarizing it and partly reducing the
field.
